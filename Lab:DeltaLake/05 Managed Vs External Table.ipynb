#### Managed Table
- Both Data and Metadata are managed by Delta Lake.
- We create Delta Table using Unity Catalog.
- UC stores data in metastore container present in Storage account registered for metastore creation in UC.

#### External Table
- Data is stored in User Specified Location
- Metadata is resides in metastore of Unity Catalog 

```sql
CREATE OR REPLACE TABLE deltacatalog.deltadb.invoices_ext
USING DELTA 
LOCATION 'abfss://labdata@dbdeltalakestorage.dfs.core.windows.net/invoices_ext'
AS
SELECT * FROM PARQUET.`abfss://labdata@dbdeltalakestorage.dfs.core.windows.net/invoices/invoices_1_100.parquet`;
```

```sql
drop table deltacatalog.deltadb.invoices_ext; 
```
