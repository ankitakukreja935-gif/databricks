```python
emp_csv = "/Volumes/samples/databricks/datasets/retail-org/company_employees/company_employees.csv"
```

```python
dbutils.fs.head(emp_csv)
```

```python
dbutils.widgets.text("dept"," ","Department Name")
```

```python
_dept = dbutils.widgets.get("dept")
```

```python
print(_dept)
```

```python
df = spark.read.csv(emp_csv,header=True)
```

```python
df_filtered = df.where(f"upper(department) = upper('{_dept}') and active_record = '1'")
```

```python
_count = df_filtered.count()
if _count > 0 :
    df_filtered.write.format("delta").mode("overwrite").option("path",f"abfss://data@dbcertificationstg.dfs.core.windows.net/easewithdata/dept_{_dept}").saveAsTable(f"dbcertification.easewithdata.dept_{_dept}")
    print(f"Data Written for dept_{_dept}")
else:
    print("No records found for the given department")
```

```python
dbutils.notebook.exit(_count)
```
