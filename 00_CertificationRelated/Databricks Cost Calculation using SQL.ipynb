- **In databricks all cost related details are stored into** `System Catalog`
- `Billing - Schema`
- Two Tables : `list_prices` and `usage`

**This cell gets latest 1000 usage records from the** `system.billing.usage` **table**, 
showing the most recent usage data for your databricks account.

```sql
Select * 
from system.billing.usage
order by usage_date desc
limit 1000
```

```sql
select usage_metadata.job_id, sku_name, count(usage_metadata.job_run_id)
from
system.billing.usage
where usage_metadata.job_id is not null 
group by usage_metadata.job_id, sku_name
```

```sql
select *
from system.billing.usage
where usage_metadata.job_run_id= '526980379285047'
order by usage_start_time desc
```

```sql
Select pricing.default
from
system.billing.list_prices
where sku_name = "PREMIUM_JOBS_SERVERLESS_COMPUTE_US_EAST"
and price_end_time is null;
```

```sql
select 
    sum(usage_quantity)
from 
    system.billing.usage
where 
    usage_metadata.job_run_id= '526980379285047'
```

```sql
select sum(usage.usage_quantity * prices.pricing.default) as Total_Cost
from system.billing.usage as usage
join
(
    select * 
    from
    system.billing.list_prices as prices
    where price_end_time is null
) as prices
on usage.sku_name = prices.sku_name
where usage.usage_metadata.job_run_id= '526980379285047'
```

```sql
Select 
usage_date,
compute_type,
SUM(cost) AS Total_Cost_USD,
SUM(cost)*95.70 AS Total_Cost_INR
FROM
(
    SELECT usage.usage_quantity * prices.pricing.default AS cost, 
    usage.sku_name AS compute_type,
    *
    FROM system.billing.usage AS usage
    JOIN
    (
        SELECT * 
        FROM
        system.billing.list_prices as prices
        where price_end_time IS NULL
    ) AS prices
    on usage.sku_name = prices.sku_name
)
group by usage_date, compute_type
order by usage_date
```
