**withColumnRenamed()** : is used to rename columns in a DataFrame. It helps improve readability and standardize column names. It is useful when integrating multiple datasets. This function does not modify original DataFrame but returns a new one.

- df = df.withColumnRenamed("oldName", "newName")

**Handling Missing Data** : Missing data can be handled using **dropna()** or **fillna()** methods. 
- dropna() removes rows containing null values. 
- fillna() replaces null values with default values. This improves data quality and prevents errors during processing.

- df.dropna().show()
- df.fillna({"age":30, "name":"Unknown"}).show()
