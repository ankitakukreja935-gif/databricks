```python
data = [
    ("North","Laptop",10),
    ("North","Phone",20),
    ("South","Laptop",5),
    ("South","Phone",15),
    ("West","Laptop",7),
    ("West","Tablet",3),
    ("East","Laptop",18)
]

df = spark.createDataFrame(data, ["region","product","qty"])
```

```python
display(df)
```

```python
display(df.groupBy("region").pivot("product").sum("qty").alias("sum_qty"))
```

```python
new_df = df.groupBy("region").pivot("product").sum("qty").alias("sum_qty")
```

```python
from pyspark.sql.functions import expr

display(new_df.select("region",expr("stack(3,'Laptop',Laptop,'Phone', Phone, 'Tablet', Tablet) as(product,qty)")))
```
