```python
from pyspark.sql import functions as F
```

```python
df = spark.createDataFrame([(1,), (2,), (3,)], ["num"])
```

```python
map_df = df.select(F.array(F.col("num"), F.col("num") * 2).alias("mapped_array"))
```

```python
display(map_df)
```

```python
flat_df = df.select(F.explode(F.array(F.col("num"), F.col("num") * 2)).alias("flattened_num"))
```

```python
display(flat_df)
```

```python
data = [
    ("Apache Spark is a unified analytics engine",),
    ("Spark provides high-level APIs in Python Java and Scala",),
    ("DataFrames optimize execution with the Catalyst optimizer",)
]
df = spark.createDataFrame(data, ["text"])
```

```python
display(df)
```

```python
words_df = (
    df
    # Clean non-alphanumeric characters and convert to lower case
    .select(F.lower(F.regexp_replace(F.col("text"), r"[^\w\s]", "")).alias("cleaned_text"))
    # Split string into an array of words
    .select(F.split(F.col("cleaned_text"), r"\s+").alias("word_array"))
    # Explode array elements into distinct rows
    .select(F.explode(F.col("word_array")).alias("word"))
    # Filter out any empty strings
    .filter(F.col("word") != "")
)
```

```python
display(df.select(F.split(F.col("text")," ").alias("word_array")))
```

```python
display(words_df)
```

```python
word_count_df = (
    words_df
    .groupBy("word")
    .agg(F.count("*").alias("count"))
    .orderBy(F.col("count").desc())
)
```

```python
display(word_count_df)
```
