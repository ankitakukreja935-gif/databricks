```python
%sql
select sum(usg.usage_quantity * prices.pricing.default)  as cost from 
 system.billing.usage as usg
 join 
 (
    select * from system.billing.list_prices
    where price_end_time is null
 ) as prices
    on usg.sku_name = prices.sku_name
    where usg.usage_metadata.job_run_id = '915036246512156'
```

```python
%sql
Select usage_date, 
        compute_type,
        custom_tags.department as department,
        SUM(cost) AS total_cost
FROM
(
select usg.usage_quantity * prices.pricing.default AS cost,
usg.sku_name AS compute_type,
*
from
 system.billing.usage as usg
 join 
 (
    select * 
    from system.billing.list_prices AS prices
    where price_end_time is null

 ) As prices
    on usg.sku_name = prices.sku_name
)
GROUP BY usage_date, compute_type, custom_tags.department
ORDER BY usage_date
```
