```python
from pyspark.sql.functions import col

data = [
    ("Product1",100,"Category1"),
    ("Product2",200,"Category2"),
    ("Product3",150,"Category1"),
    ("Product4",300,"Category3")
]

columns = ["ProductName","Price","Category"]

df = spark.createDataFrame(data, columns)

df.display()
```

```python
from pyspark.sql.functions import transform

def transform_function(df):
    return df.withColumn("DiscountedPrice", col("Price")* 0.9)

# Apply Transform function using transform method

transformed_df = df.transform(transform_function)

transformed_df.display()
```

```python
from pyspark.sql.functions import col

data = [
    ("Product1",100,"Category1"),
    ("Product2",200,"Category2"),
    ("Product3",150,"Category1"),
    ("Product4",300,"Category3")
]

columns = ["ProductName","Price","Category"]

df = spark.createDataFrame(data, columns)

df.display()
```

```python
def transform_function(df, discount_percentage):
    return df.withColumn("DiscountedPrice", col("Price") * (1 - discount_percentage/100))


percentage = 20

# Different Approaches to call function 
transformed_df1 = df.transform(transform_function,percentage)
```

```python
display(transformed_df1)
```

```python
percentage = 10
transformed_df2 = df.transform(transform_function, discount_percentage = percentage)
display(transformed_df2)
```

```python
percentage = 10
transformed_df3 = df.transform(lambda df: transform_function(df, percentage))
display(transformed_df3)
```
