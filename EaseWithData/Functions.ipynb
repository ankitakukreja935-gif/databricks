```sql
-- SCALER FUNCTION FOR API
CREATE OR REPLACE FUNCTION dbcertification.easewithdata.fruit_nutrition(fruit_name STRING)
RETURNS STRING
LANGUAGE PYTHON
AS
$$
    import requests
    import json

    api_url = f"https://www.fruityvice.com/api/fruit/{fruit_name}"
    response = requests.get(api_url)
    data = response.json()
    return str(data.get('nutritions','NA'))
$$
;
```

```sql
CREATE OR REPLACE FUNCTION dbcertification.easewithdata.fruit_nutrition_new(fruit_name STRING)
RETURNS STRING
LANGUAGE PYTHON
AS
$$
    import requests
    import json

    if not fruit_name:
        return "NA"

    try:
        api_url = f"https://www.fruityvice.com/api/fruit/{fruit_name.strip()}"
        response = requests.get(api_url, timeout=5)
        
        if response.status_code == 200:
            data = response.json()
            nutritions = data.get("nutritions")
            return json.dumps(nutritions) if nutritions else "NA"
        else:
            return "NA"
    except Exception:
        return "NA"
$$;
```

```sql
select
    v.* from
    (
select from_json(dbcertification.easewithdata.fruit_nutrition("banana"),'calories double, fat double, sugar double, carbohydrates double, protein double') as v
    )
```

```sql
select
    v.* from
    (
select from_json(dbcertification.easewithdata.fruit_nutrition_new("guava"),'calories double, fat double, sugar double, carbohydrates double, protein double') as v
    )
```
